7. Sales And Profit Analysis 

Category wise sales

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending = False)
)
category_sales
```

Category wise profit



```python
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending = False)
)
category_profit
```

Sub-Category wise profit

```python
sub_category_profit = (
    df.groupby("Sub-Category")["Profit"]
    .sum()
    .sort_values(ascending = False)
)
sub_category_profit
```

Region wise profit

```python
region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending = False)
)
region_profit
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data = df,
    x = "Category",
    y = "Sales",
    estimator = "sum"
)

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

Category wise profit 

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data = df,
    x = "Category",
    y = "Profit",
    estimator = "sum"
)

plt.title("Category-wise Total profit")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()
```

Sales distribution 

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data = df,
    y = "Sales"
)

plt.title("Sales Distribution")
plt.ylabel("Sales")

plt.show()
```

Profit Box Plot 

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data = df,
    y = "Profit"
)

plt.title("Profit Distribution")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data = df,
    x = "Category",
    y = "Profit"
)

plt.title("Profit Distribution")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data = df,
    x = "Discount",
    y = "Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

Calculating Profit Margin

```python
df["Profit Margin"] = (df["Profit"]/df["Sales"])*100

df[["Sales","Profit","Profit Margin"]].head()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data = df,
    x = "Discount",
    y = "Profit Margin"

)

plt.title("Discount vs Profit Margin")
plt.xlabel("Discount")
plt.ylabel("Profit Margin (%)")

plt.show()
```

Correlation Analysis

```python
numerical_df = df.select_dtypes(include = np.number)

correlation_matrix = numerical_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation_matrix,
    annot = True,
    fmt = ".2f",
    cmap = "coolwarm"
)

plt.title("Correlation Heatmap of Numerical Attributes")


plt.show()
```
